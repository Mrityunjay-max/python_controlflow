```python
amount = int(input("Enter order amount:"))
if amount >= 2000:
    Discount = (amount*15)/100
    final_amount = amount - Discount

elif amount >= 1500:
    Discount = (amount*10)/100
    final_amount = amount - Discount
elif amount >= 1000:
    Discount = (amount*7)/100
    final_amount = amount - Discount
else :
    print("no discount")
    final_amount = amount

print("Final amount is ",final_amount)
```

```python
orders = [1200,2500,800,1750,3000]
final_amount = []
for i in orders:
    if i >= 2000:
        Discount = (i*15)/100
        final_amount.append(i-Discount)
    elif i >=1500:
        Discount = (i*10)/100
        final_amount.append(i-Discount)
    elif i>=1000:
        Discount = (i*7)/100
        final_amount.append(i-Discount)
    else :
        final_amount.append(i)
        
print(final_amount)
TotalAmount = sum(final_amount)
print("Total revenue",TotalAmount)
```
